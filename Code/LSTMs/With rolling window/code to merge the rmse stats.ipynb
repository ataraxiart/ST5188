```python
import pandas as pd
```

```python
# Define window sizes
window_sizes = range(9, 14)

# Create an empty list to hold the DataFrames
rmse_dfs = []

for window in window_sizes:
    df = pd.read_csv(f"rmse_{window}.csv", index_col="Forecast horizon")
    df.columns = [f"Window = {window}"]
    rmse_dfs.append(df)

# Concatenate along columns
merged_rmse_df = pd.concat(rmse_dfs, axis=1)

# Save to new CSV
merged_rmse_df.to_csv("merged_rmse.csv")

print(merged_rmse_df)
```

```python
# Define window sizes
window_sizes = range(9, 14)

# Create an empty list to hold the DataFrames
rmse_dropout_dfs = []

for window in window_sizes:
    df = pd.read_csv(f"rmse_dropout_{window}.csv", index_col="Forecast horizon")
    df.columns = [f"Window = {window}"]
    rmse_dropout_dfs.append(df)

# Concatenate along columns
merged_rmse_dropout_df = pd.concat(rmse_dropout_dfs, axis=1)

# Save to new CSV
merged_rmse_dropout_df.to_csv("merged_rmse_dropout.csv")

print(merged_rmse_dropout_df)
```
