```python
# imports
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

### Training and test splitting

#### Changi

```python
df = pd.read_csv("../../Data/Final/final_CHANGI_wide.csv")
```

```python
train_df, test_df = train_test_split(df, test_size=0.2, random_state=42) # 80-20 split

# wide format
train_df.to_csv("../../Data/Final/TT Split/CHANGI_train_wide.csv", index=False)
test_df.to_csv("../../Data/Final/TT Split/CHANGI_test_wide.csv", index=False)

# convert wide format into long format
train_long_df = pd.melt(train_df, id_vars=['x', 'y'], var_name='Date', value_name='Value')
test_long_df = pd.melt(test_df, id_vars=['x', 'y'], var_name='Date', value_name='Value')

train_long_df.to_csv("../../Data/Final/TT Split/CHANGI_train_long.csv", index=False)
test_long_df.to_csv("../../Data/Final/TT Split/CHANGI_test_long.csv", index=False)
```
