### Training and test splitting

#### Changi

```python
changi_df = pd.read_csv("../../Data/Final/final_CHANGI_wide.csv")
changi_df.info
```

```python
start_col = "Mar-Apr 2000"
end_col = "Sep-Oct 2022"

end_idx = changi_df.columns.get_loc(end_col)

# Train Set: Includes up to Nov-Dec 2019
changi_train = changi_df.loc[:, ["x", "y"]].join(changi_df.loc[:, start_col:end_col])

# Test Set: Everything AFTER Nov-Dec 2019 (excluding it)
changi_test = changi_df.loc[:, ["x", "y"]].join(changi_df.iloc[:, end_idx + 1:])

# Wide format
changi_train.to_csv("../../Data/Final/TT Split/changi_train_wide.csv", index=False)
changi_test.to_csv("../../Data/Final/TT Split/changi_test_wide.csv", index=False)

# Long format
changi_train_long = changi_train.melt(id_vars=["x", "y"], var_name="Date", value_name="Value")
changi_test_long = changi_test.melt(id_vars=["x", "y"], var_name="Date", value_name="Value")

changi_train_long.to_csv("../../Data/Final/TT Split/changi_train_long.csv", index=False)
changi_test_long.to_csv("../../Data/Final/TT Split/changi_test_long.csv", index=False)

print("Train and test sets saved successfully!")
print(changi_test.info())
```

#### Training test split for Imputed Jurong West (for ARIMA/LSTM)

```python
df = pd.read_csv("../../Data/Final/Imputation/changi_imp_final.csv")
print(df.shape)
```

```python
# Function to extract the year from the 'Date' column
def extract_year(date_str):
    # The date is in the format "Mar-Apr 2000", so we can split by the space
    year = date_str.split()[-1]
    return int(year)

# Apply the function to create a new 'Year' column
df['Year'] = df['Date'].apply(extract_year)

# Define the training set (Year <= 2022) and the test set (Year > 2022)
train_set = df[df['Year'] <= 2022]
test_set = df[df['Year'] > 2022]

train_set = train_set.drop(columns=['Year'])
test_set = test_set.drop(columns=['Year'])

# Output the shapes of the training and test sets to confirm the split
print(f"Training Set Shape: {train_set.shape}")
print(f"Test Set Shape: {test_set.shape}")

# Optionally, save the training and test sets to separate CSV files
train_set.to_csv('../../Data/Final/Imputation/imp_train_set.csv', index=False)
test_set.to_csv('../../Data/Final/Imputation/imp_test_set.csv', index=False)
```
